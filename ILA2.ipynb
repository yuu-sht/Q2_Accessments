```python
secret_number = 17
attempts = 0

while attempts < 5:
  guess =int(input("Enter your guess:"))

  if guess < 0:
    continue

  attempts += 1

  if guess == secret_number:
    break

if attempts == 5 and guess != secret_number:
  print("Game Over!")
```
