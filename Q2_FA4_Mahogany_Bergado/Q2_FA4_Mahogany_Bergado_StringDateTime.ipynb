## 🧰 Part 1: Predefined Libraries & String Manipulation

A **Library** is like a toolbox. If a useful tool already exists, you don't need to build it from zero!

### Common String Cleaning Tasks:
* `strip()`: Removes extra leading and trailing spaces.
* `title()`: Capitalizes the first letter of each word (proper name formatting).
* `upper()`: Converts all characters to UPPERCASE.
* `lower()`: Converts all characters to lowercase.
* `len()`: Counts total characters (including spaces).
* `find(substr)`: Locates the starting position of a substring.
* `isdigit()` / `isalpha()`: Checks if text consists strictly of numbers or letters.



````python
# Run this cell to test Python String Cleaning Methods!

messy_name = "   juan dela cruz   "
messy_section = " magnesium "
messy_subject = "computer science"

# 1. Clean extra spaces & format proper title case
clean_name = messy_name.strip().title()
print("Cleaned Name:", clean_name)

# 2. Convert to Uppercase
name_code = clean_name.upper()
section_code = messy_section.strip().upper()
print("Name Code:", name_code)
print("Section Code:", section_code)

# 3. Format Subject & Count Characters
clean_subject = messy_subject.strip().title()
char_count = len(clean_name)

print("Formatted Subject:", clean_subject)
print("Character Count of Name:", char_count)

````

## ⏰ Part 2: Predefined Libraries for Date & Time

Dates are tricky because of leap years, month length differences, and different global formats. Python's `datetime` module makes handling schedules easy!

### Key `strftime()` Formatting Codes:
* `%B` : Full month name (e.g., `June`)
* `%b` : Abbreviated month name (e.g., `Jun`)
* `%d` : Day of the month with zero-padding (e.g., `03`)
* `%Y` : 4-digit year (e.g., `2026`)
* `%A` : Full weekday name (e.g., `Wednesday`)



````python
from datetime import datetime

# Creating a specific date object: datetime(year, month, day)
event_date = datetime(2026, 6, 3)

# Formatting using strftime()
readable_date = event_date.strftime("%B %d, %Y")
weekday_name = event_date.strftime("%A")
month_name = event_date.strftime("%B")

print("Formatted Date:", readable_date)
print("Year:", event_date.year)
print("Month:", month_name)
print("Day:", event_date.day)
print("Weekday:", weekday_name)

````

## 📅 Activity 2: Deadline & Date Formatter

**Task:** Prompt the user for an event title, year, month, and day, then output a clean schedule card with the weekday!



````python
# Verion 1: Basic input and output without date formatting

# Input Data from user
from datetime import datetime

event_title = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

# Compute weekday
weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
weekday = weekdays[datetime(year, month, day).weekday()]

# Create datetime object
event_date = datetime(year, month, day)

#Display output
print("\n --- Formatted Output ---")
print(f"Event: {event_title.upper()}")
print(f"Date: {month}/{day}/{year}")
print(f"Year: {year}")
print(f"Month: {month}")
print(f"Day: {day}")
print(f"Weekday: {weekday}")

````

````python
# Version 2: Added Date Formatting

# Input Data from user
from datetime import datetime

event_title = input("Enter event title: ").strip()
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

# Create datetime object
event_date = datetime(year, month, day)

# Format date strings using strftime()
formatted_date = event_date.strftime("%B %-d, %Y")
month_name = event_date.strftime("%B")
weekday_name = event_date.strftime("%A")

# Output Data
print("\n--- Formatted Output ---")
print(f"Event: {event_title.upper()}")
print(f"Date: {formatted_date}")
print(f"Year: {event_date.year}")
print(f"Month: {month_name}")
print(f"Day: {event_date.day}")
print(f"Weekday: {weekday_name}")
````

## 🔄 Part 3: Integrating Strings, Dates, Arrays & Loops

Real-world applications often process lists of records. Below, we loop over an array of unformatted student registrations, cleaning the text and formatting the dates automatically.



````python

from datetime import datetime

# Parallel arrays for student registration details
names = ["   juan dela cruz ", "maria  clara ","  pedro penduko"]

sections = ["molave ", " acacia", "Mahogany"]

events = [" science fair ", "  robotics competition ", "hackathon "]

years = [2026, 2026, 2027]
months = [6, 11, 2]
days = [3, 15, 20]

print("=== AUTOMATED BATCH REGISTRATION FORMATTER ===")

# Process each registration using the same index
for idx in range(len(names)):

    # Clean text
    clean_name = names[idx].strip().title()
    name_code = clean_name.upper()
    clean_section = sections[idx].strip().upper()
    clean_event = events[idx].strip().title()

    # Format date
    event_dt = datetime(years[idx], months[idx], days[idx])
    formatted_date = event_dt.strftime("%B %-d, %Y")

    # Display formatted registration
    print(f"\nName: {clean_name}")
    print(f"Name Code: {name_code}")
    print(f"Section: {clean_section}")
    print(f"Event: {clean_event}")
    print(f"Date: {formatted_date}")
````

## 🚀 Part 4: Independent Learning Activity-Student Event Registration System

### Complete Application Implementation with GitHub Versioning

Follow the study guide's versioning requirements:
* **Version 1:** Basic input and raw output.
* **Version 2:** Added string formatting (`strip()`, `title()`, `upper()`).
* **Version 3:** Added date/time formatting (`datetime`, `strftime()`).



````python
#  Version 3 Implementation (Student Event Registration Formatter)
from datetime import datetime

def run_event_registration():
    print("=== Student Event Registration Formatter ===")

    # 1. Inputs
    fname = input("Student First Name: ")
    lname = input("Student Last Name: ")
    sec = input("Section: ")
    evt = input("Event Name: ")
    yr = int(input("Event Year (YYYY): "))
    mo = int(input("Event Month (1-12): "))
    dy = int(input("Event Day (1-31): "))

    # 2. String Manipulation
    full_name = f"{fname.strip()} {lname.strip()}".title()
    name_code = full_name.upper()
    section_upper = sec.strip().upper()
    event_clean = evt.strip().title()


    # 3. Date Handling
    event_dt = datetime(yr, mo, dy)
    date_formatted = event_dt.strftime("%B %-d, %Y")

    # 4. Formatted Summary Output
    print("" + "=" *40)
    print("  Student Event Registration Summary  ")
    print("=" *40)
    print(f"Name                  : {full_name}")
    print(f"Name Code             : {name_code}")
    print(f"Section               : {section_upper}")
    print(f"Event                 : {event_clean}")
    print(f"Event Date            : {date_formatted}")
    print(f"Language Used         : Python")
    print(f"String Feature Used   : strip(), title(), upper()")
    print(f"Date/Time Feature Used: datetime, strftime()")
    print("="*40)
    print("Registration record created successfully.")

# Execute interactive registration
run_event_registration()

````
