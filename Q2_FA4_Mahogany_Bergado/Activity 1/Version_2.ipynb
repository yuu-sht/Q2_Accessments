```python
# Version 2: Added string formatting

# Get clean inputs by removing extra leading/trailing space
first_name = input("Enter first name: ").strip()
last_name = input("Enter last name: ").strip()
section = input("Enter section: ").strip()
favorite_subject = input("Enter favorite subject: ").strip()

#Format full name in proper cases and its character count
full_name = f"{first_name} {last_name}".title()
char_count = len(full_name)

#Diplay formatted outputs
print("\n --- Formatted Output ---")
print(f"Name: {full_name}")
print(f"Name Code: {full_name.upper()}")
print(f"Section: {section.upper()}")
print(f"Favorite Subject: {favorite_subject.title()}")
print(f"Character Count: {char_count}")
```
